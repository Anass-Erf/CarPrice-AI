```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time 
```

```python
base_url = 'https://www.wandaloo.com/neuf/maroc/0,0,0,0,0,0,-,az,'
datas = []
for page in range(1, 60):  # Adjust the range for more pages
    print(f"Scraping page {page}...")
    url = base_url + str(page)+".html"

    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }

    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        page_content = response.text
    else:
        print(f'Failed to retrieve the page. Status code: {response.status_code}')    
    soup = BeautifulSoup(page_content, 'html.parser')
    
    car_models = soup.find_all('li')  # Replace with actual class
    if not car_models:
        print("No more car listings found. Stopping.")
        
    for model in car_models:
        model_Name = model.find('h3' , class_="titre" )
        price_Intervall = model.find('p', class_="prix")
        
        model_Name = model_Name.text.strip() if model_Name else "N/A"
        price_Intervall = price_Intervall.text.strip() if price_Intervall else "N/A"
        # datas.append({'MODEL': model_Name, 'INTERVALE': price_Intervall,})# 'VERSION': Version, 'VERSION_PRICE': Version_price})
    
        car_versions = model.find_all('li', class_="item")
        
        for version in car_versions:
            Version = version.find('a')
            Version_price = version.find('li', class_="prix")
            Version = Version.text.strip() if Version else "N/A"
            Version_price = Version_price.text.strip() if Version_price else "N/A"
            datas.append({'MODEL': model_Name, 'INTERVALE': price_Intervall, 'VERSION': Version, 'VERSION_PRICE': Version_price})
    time.sleep(2)  # Add delay to prevent blocking

for data in datas:
    print(data)
```

```python
df = pd.DataFrame(datas)
df.to_csv("car_listings_3.csv", index=False, encoding="utf-8")

print(f"Scraping completed. Data saved in 'car_listings_3.csv'.")
```
